```python
#3 Analisando uma Tupla
#Crie uma tupla com 6 números inteiros. Exiba a quantidade de números pares, a soma dos números ímpares e o menor valor armazenado.

numeros = (12,21,25,15,5,6)
menor = numeros[0]
soma = 0
pares = 0
for numero in numeros:
    #Procura os pares
    if numero%2 == 0:
        pares+=1
    else:
    #faz a soma dos ímpares da Tupla
        soma+=numero
    #Procura o menor valor
    if numero < menor:
        menor = numero

print(f'O menor valor é: {menor}\nA quantidade de pares é: {pares}\nA soma dos valores ímpares da tupla é {soma}.')
```

```python
#7 Notas Imutáveis
#Crie uma tupla com quatro notas de um aluno. Exiba a média das notas, a maior nota e a menor nota. Mostre também todas as notas armazenadas.

notas = (10.0,9.5,8.5,9.0)
soma = 0
maior = max(notas)
menor = min(notas)
for nota in notas:
    soma += nota
media = soma/len(notas)
print(f'Média de Notas: {media}, Maior Nota: {maior}, Menor Nota: {menor}')
```

```python
#12 Classificação de Temperaturas
#Crie uma tupla com temperaturas registradas durante a semana. Exiba quantas temperaturas estão acima da média e informe a maior e a menor temperatura registrada.

temperaturas = (35.6,31.4,25.6,21.4,23.6,26.4,27.5)
media = sum(temperaturas) / len(temperaturas)
maior = max(temperaturas)
menor = min(temperaturas)
print(f'Temperatura Média: {media:.1f}°C\nTemperatura Máxima: {maior}°C\nTemperatura Mínima: {menor}°C')
```

```python
#14 Posições em Tupla
#Crie uma tupla com 8 números inteiros. Leia um número e verifique em quais posições ele aparece dentro da tupla. Caso não exista, exiba uma mensagem apropriada.
numeros = (2,2,3,3,2,4,5,6)
verificador = int(input('Digite sua busca: '))
if verificador not in numeros:
    print(f'O número {verificador} não está na Tupla.')
else:
    for posicao in range(len(numeros)):
        if verificador ==  numeros[posicao]:
            print(f'O número {verificador} está na Posição: {posicao}')
```

```python
#17 Comparando Conjuntos
#Crie dois conjuntos de números inteiros. Verifique se os conjuntos são iguais, diferentes ou se um está contido dentro do outro. Exiba mensagens apropriadas para cada caso.

conjuntoA = {2,4,5,6,7,9}
conjuntoB = {5,7,9}
if conjuntoA == conjuntoB:
    print('Os conjuntos são iguais')
else:
    print('Os conjuntos são diferentes')

if conjuntoB.issubset(conjuntoA):
    print('O conjunto B está contido em A')
else:
    print('O conjunto B não está contido em A')

if conjuntoA.issubset(conjuntoB):
    print('O conjunto A está contido em B')
else:
    print('O conjunto A não está contido em B')
```

```python
#20 Resumo Estatístico de Tupla
#Crie uma tupla com 10 números inteiros. Exiba a soma total, a média dos valores, o maior número, o menor número e quantos valores são pares.

num_int = (951,242,578,154,147,124,547,211,21,10)
pares = 0
for num in num_int:
    if num%2 == 0:
        pares+=1
maior = max(num_int)
menor = min(num_int)
soma = sum(num_int)
media = soma/len(num_int)

print(f'Soma Total: {soma}\nMédia: {media}\nMaior valor: {maior}\nMenor valor:{menor}\nPares: {pares}')
```
