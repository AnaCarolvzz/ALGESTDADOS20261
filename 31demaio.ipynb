```python
# Leitura do conjunto A
conjuntoA = set()

print("Digite os 10 elementos do conjunto A:")

for i in range(10):
    valor = input("Elemento: ")
    conjuntoA.add(valor)

# Leitura do conjunto B
conjuntoB = set()

print("\nDigite os 10 elementos do conjunto B:")

for i in range(10):
    valor = input("Elemento: ")
    conjuntoB.add(valor)

# Operações com conjuntos
intersecao = conjuntoA & conjuntoB
uniao = conjuntoA | conjuntoB
diferencaAB = conjuntoA - conjuntoB
diferencaBA = conjuntoB - conjuntoA

# Exibição dos resultados
print("\nConjunto A:", conjuntoA)
print("Conjunto B:", conjuntoB)

print("\nInterseção:", intersecao)
print("União:", uniao)
print("Diferença A - B:", diferencaAB)
print("Diferença B - A:", diferencaBA)
```

```python
frase = input("Digite uma frase: ")

letras = set()

for caractere in frase:
    if caractere != " " and caractere != "." and caractere != "," and caractere != "!" and caractere != "?" and caractere != ";" and caractere != ":":
        letras.add(caractere)

print("Letras utilizadas na frase:")
print(letras)
```

```python
numeros = set()

for i in range(1, 101):
    if i % 2 != 0 and i % 3 != 0:
        numeros.add(i)

print("Números que não são múltiplos de 2 nem de 3:")
print(numeros)
```

```python
conjuntoA = set()
conjuntoB = set()

print("Digite 3 números para o conjunto A:")

for i in range(3):
    numero = int(input("Número: "))
    conjuntoA.add(numero)

print("\nDigite 5 números para o conjunto B:")

for i in range(5):
    numero = int(input("Número: "))
    conjuntoB.add(numero)

if conjuntoA.issubset(conjuntoB):
    print("\nO conjunto A é subconjunto de B.")
elif conjuntoB.issubset(conjuntoA):
    print("\nO conjunto B é subconjunto de A.")
else:
    print("\nNenhum conjunto é subconjunto do outro.")
```

```python
#24 Dado um conjunto de números inteiros, crie um novo conjunto com apenas os quadrados perfeitos encontrados.
numeros = set()

print("Digite 10 números inteiros:")

for i in range(10):
    numero = int(input("Número: "))
    numeros.add(numero)

quadrados_perfeitos = set()

for numero in numeros:

    for i in range(numero + 1):

        if i * i == numero:
            quadrados_perfeitos.add(numero)

print("\nConjunto original:")
print(numeros)

print("\nQuadrados perfeitos encontrados:")
print(quadrados_perfeitos)
```

```python
funcionarios = {}

# Entrada de dados
for i in range(5):
    nome = input("Digite o nome do funcionário: ")
    salario = float(input("Digite o salário: "))

    funcionarios[nome] = salario

# Reajuste de salários
for nome in funcionarios:
    if funcionarios[nome] < 2000:
        funcionarios[nome] = funcionarios[nome] * 1.10

# Saída
print("\nSalários atualizados:")
for nome in funcionarios:
    print(nome, "-> R$", funcionarios[nome])
```

```python
alunos = {}

# Entrada de dados
for i in range(5):
    nome = input("Digite o nome do aluno: ")

    notas = []

    for j in range(3):
        nota = float(input("Digite uma nota: "))
        notas.append(nota)

    alunos[nome] = notas

# Cálculo e saída
print("\nMédia dos alunos:")

for nome in alunos:
    soma = 0

    for nota in alunos[nome]:
        soma = soma + nota

    media = soma / 3

    print(nome, "-> Média:", media)
```

```python
palavras = {}

for i in range(10):
    palavra = input("Digite uma palavra: ")

    letra = palavra[0]

    if letra in palavras:
        palavras[letra].append(palavra)
    else:
        palavras[letra] = [palavra]

print("\nDicionário organizado por primeira letra:")
print(palavras)
```

```python
produtos = {}

for i in range(3):
    nome = input("Digite o nome do produto: ")

    precos = []

    for j in range(3):
        preco = float(input("Digite o preço na loja: "))
        precos.append(preco)

    produtos[nome] = precos

maior_variacao = -1
produto_maior_variacao = ""

for produto in produtos:
    precos = produtos[produto]

    maior_preco = precos[0]
    menor_preco = precos[0]

    for preco in precos:
        if preco > maior_preco:
            maior_preco = preco
        if preco < menor_preco:
            menor_preco = preco

    variacao = maior_preco - menor_preco

    if variacao > maior_variacao:
        maior_variacao = variacao
        produto_maior_variacao = produto

print("\nProduto com maior variação de preço:")
print(produto_maior_variacao, "-> Variação:", maior_variacao)
```

```python
frase = input("Digite uma frase: ")

frase = frase.lower()

palavras = frase.split()

contagem = {}

for palavra in palavras:
    if palavra in contagem:
        contagem[palavra] = contagem[palavra] + 1
    else:
        contagem[palavra] = 1

print("\nContagem de palavras:")
print(contagem)
```
