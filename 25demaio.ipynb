```python
#1 Crie a ficha de um produto
#Crie um dicionário chamado produto com as chaves: nome, preco, estoque e categoria. Imprima cada campo com for e .items().
#Depois, atualize o preço e adicione uma chave desconto com um valor de sua escolha.

produto = {
    'nome':'arroz sepe 5kg',
    'perco': 14.98,
    'estoque': 200,
    'categoria': 'pereciveis'}

for descição in produto:
    print(f'{descição}: {produto[descição]}')

print(produto.items())

```

```python
#2 Cadastro de múltiplos alunos
#Crie uma lista onde cada item é um dicionário representando um aluno (com nome e media).
#Use um for para percorrer a lista e imprimir: 'Aluno: [nome] — Média: [media]'. Depois,
#imprima apenas os alunos com média acima de 7.

aluno1 = {'nome': 'carol', 'media': 7.5}
aluno2 = {'nome': 'henriqule', 'media': 8.0}
aluno3 = {'nome': 'bahia', 'media': 9.5}
aluno4 = {'nome': 'hermes', 'media': 6.5}
aluno5 = {'nome': 'tailine', 'media': 9.0}

alunos = [aluno1, aluno2, aluno3, aluno4, aluno5]

for aluno in alunos:
    print(f'aluno: {aluno}')

print('\nalunos acima da media 7.0:')

for aluno in alunos:
    if aluno['media']>=7:
      print(f'aluno: {aluno['nome']} - media: {aluno['media']}')
```

```python
#3 Calculando valor total em estoque
#Crie um dicionário chamado produto com as chaves nome, preco e quantidade; em seguida, calcule
#o valor total em estoque multiplicando o preço pela quantidade e armazene esse resultado em uma
#nova chave chamada valor_total; utilize um for com .items() para exibir todas as informações do
#roduto, incluindo o valor total calculado.

produto = {
    'nome': 'iphone 13',
    'preco': 6500.00,
    'quantidade': 15
}

# Calcula o valor total em estoque
produto['valor_total'] = produto['preco'] * produto['quantidade']
```
